# 🖥️ colab-desktop: Full Linux Desktop (XFCE4) + Native Firefox in Google Colab

[![GitHub](https://img.shields.io/badge/GitHub-envblex%2Fcolab--desktop-blue?logo=github)](https://github.com/envblex/colab-desktop)
[![License: MIT](https://img.shields.io/badge/License-MIT-yellow.svg)](https://opensource.org/licenses/MIT)

Run a complete Debian/Ubuntu XFCE4 desktop environment directly inside your Google Colab notebook.
- **Works on Free CPU & GPU Runtimes**
- **Interactive Web Desktop (noVNC)** embedded right below this cell
- **Official Native Mozilla Firefox** (non-snap, full WebRTC & hardware accelerated)
- **Zero configuration**: Click the play button below and start browsing in 60 seconds.

```python
#@title 🚀 Run Desktop in Colab { run: "auto" }
#@markdown Select your preferred desktop resolution and browser options:
RESOLUTION = "1280x720" #@param ["1280x720", "1920x1080", "1600x900", "1024x768"]
AUTO_LAUNCH_FIREFOX = True #@param {type:"boolean"}
EMBED_WINDOW_HEIGHT = 750 #@param {type:"integer"}

import subprocess
from IPython.display import display, HTML
from google.colab.output import serve_kernel_port_as_iframe, serve_kernel_port_as_window

# Execute automated setup script from GitHub
setup_url = "https://raw.githubusercontent.com/envblex/colab-desktop/main/scripts/setup.sh"
cmd = f"curl -fsSL '{setup_url}' | bash -s -- '{RESOLUTION}' '{str(AUTO_LAUNCH_FIREFOX).lower()}'"
subprocess.run(cmd, shell=True)

# Standalone browser window link
serve_kernel_port_as_window(6080)

# Embed desktop window directly into Colab output frame
display(HTML("<h3>👇 Interactive Linux Desktop:</h3>"))
serve_kernel_port_as_iframe(6080, width="100%", height=str(EMBED_WINDOW_HEIGHT))
```
